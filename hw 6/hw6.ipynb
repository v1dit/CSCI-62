```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import preprocessing
```

```python
data = pd.read_csv("clean_data.csv")
```

```python
data.head()
```

```python
data.shape
```

```python
data.isnull().sum()
```

```python
threshold = len(data) * 0.35
data = data.dropna(thresh=len(data) - threshold, axis=1)
```

```python
data.columns
```

```python
bots = data[data['bot'] == 1]
nonbots = data[data['bot'] == 0]
```

```python
plt.figure(figsize=(8,6))

plt.scatter(bots['friends_count'], bots['followers_count'], 
            color='red', label='Bots', alpha=0.5)

plt.scatter(nonbots['friends_count'], nonbots['followers_count'], 
            color='blue', label='Non-Bots', alpha=0.5)

plt.xlabel("Friends Count")
plt.ylabel("Followers Count")
plt.title("Friends vs Followers")
plt.legend()

plt.show()
```

```python
data[['friends_count','followers_count','bot']].head()
```

```python
bag_of_words = ['bot', 'follow me', 'free', 'fake', 'virus', 'troll', 'clone', 'tweet me']
```

```python
def contains_words(text):
    if pd.isna(text):
        return False
    text = text.lower()
    for word in bag_of_words:
        if word in text:
            return True
    return False
```

```python
data['screen_name_binary'] = data['screen_name'].apply(contains_words)
data['name_binary'] = data['name'].apply(contains_words)
data['description_binary'] = data['description'].apply(contains_words)
data['status_binary'] = data['status'].apply(contains_words)
```

```python
data[['screen_name','screen_name_binary']].head()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.tree import export_text
```

```python
features = data[['friends_count',
                 'followers_count',
                 'screen_name_binary',
                 'name_binary',
                 'status_binary',
                 'verified',
                 'statuses_count']]

target = data['bot']
```

```python
X = features
Y = target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.3, random_state=0)
```

```python
clf = DecisionTreeClassifier(criterion='entropy')
clf.fit(X_train, y_train)
```

```python
train_pred = clf.predict(X_train)
print("Training Accuracy:", accuracy_score(y_train, train_pred))
```

```python
test_pred = clf.predict(X_test)
print("Testing Accuracy:", accuracy_score(y_test, test_pred))
```

```python
print(export_text(clf, feature_names=list(X.columns)))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
```

```python
train_pred_lr = lr.predict(X_train)
print("Logistic Regression Training Accuracy:", accuracy_score(y_train, train_pred_lr))
```

```python
train_pred_lr = lr.predict(X_train)
print("Logistic Regression Training Accuracy:", accuracy_score(y_train, train_pred_lr))
```

```python
test_pred_lr = lr.predict(X_test)
print("Logistic Regression Testing Accuracy:", accuracy_score(y_test, test_pred_lr))
```

```python
import random
import pandas as pd

X1 = []
X2 = []
Y = []

for i in range(1000):
    x1 = random.random()
    x2 = random.random()
    
    y = 1 if (x1 + x2) > 1 else 0
    
    X1.append(x1)
    X2.append(x2)
    Y.append(y)

df = pd.DataFrame({
    'X1': X1,
    'X2': X2,
    'Y': Y
})

df.head()
```

```python
import matplotlib.pyplot as plt

class1 = df[df['Y'] == 1]
class0 = df[df['Y'] == 0]

plt.scatter(class1['X1'], class1['X2'], color='red', label='Y=1')
plt.scatter(class0['X1'], class0['X2'], color='blue', label='Y=0')

plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Artificial Dataset")
plt.legend()

plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
X = df[['X1','X2']]
y = df['Y']
```

```python
tree = DecisionTreeClassifier(max_depth=2)
tree.fit(X, y)

pred_tree = tree.predict(X)

print("Decision Tree Accuracy:", accuracy_score(y, pred_tree))
```

```python
lr = LogisticRegression()
lr.fit(X, y)

pred_lr = lr.predict(X)

print("Logistic Regression Accuracy:", accuracy_score(y, pred_lr))
```

```python
Nearly 100% was expected because the rule Y = 1 if X1 + X2 > 1, and 0 otherwise
The rule defines a linear decision boundary, which is X1 + X2 = 1
Logistic Regression is designed to learn linear decision boundaries
This makes sure that it can closely approximate the true boundary in the data and classify the points correctly
```

```python
max_depth = 2 , approximately 0.865 (86.5%)
the tree is restricted to only two levels of depth so it can't capture the diagonal boundary that separates the classes
this leads to a lower accuracy
```

```python
tree_full = DecisionTreeClassifier()
tree_full.fit(X, y)
pred_full = tree_full.predict(X)
print("Decision Tree (no depth limit) Accuracy:", accuracy_score(y, pred_full))
```

```python
It can achieve near-perfect accuracy because it can split the feature space into smaller regionss until it memorizes the training data
Instead of learning the true diagonal boundary X1 + X2 = 1, the tree approximates the boundary using many small axis-aligned splits.
this way, the tree can classify almost every training point correctly
```
